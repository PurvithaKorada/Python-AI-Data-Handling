```python
import numpy as np

# Creating NumPy arrays
arr1 = np.array([10, 20, 30, 40, 50])
arr2 = np.array([[1, 2, 3], [4, 5, 6]])

print("1D Array:")
print(arr1)

print("\n2D Array:")
print(arr2)

# Array indexing
print("\nFirst element:", arr1[0])
print("Third element:", arr1[2])

# Array slicing
print("\nSliced Array:", arr1[1:4])

# Shape and dimensions
print("\nShape of arr1:", arr1.shape)
print("Shape of arr2:", arr2.shape)
print("Dimensions of arr1:", arr1.ndim)
print("Dimensions of arr2:", arr2.ndim)

# Mathematical operations
print("\nAddition:", arr1 + 5)
print("Multiplication:", arr1 * 2)

# Statistical operations
print("\nMean:", np.mean(arr1))
print("Maximum:", np.max(arr1))
print("Minimum:", np.min(arr1))
print("Sum:", np.sum(arr1))
```
