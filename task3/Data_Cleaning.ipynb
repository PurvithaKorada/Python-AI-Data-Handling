```python
import pandas as pd
# Creating a dataset with missing values,duplicate records and incorrect data types
data = {
    'Name': ['Aarav', 'Priya', 'Rahul', 'Sneha', 'Arjun', 'Aarav'],
    'Age': ['20', '21', 'nineteen', '22', '20', '20'],
    'Marks': ['85', None, '67', '95', '74', '85'],
    'Subject': ['Python', 'AI', 'Python', 'ML', 'AI', 'Python']
}
df = pd.DataFrame(data)
# Save the original dataset
df.to_csv("student_data_before_cleaning.csv", index=False)
print("Original Dataset:")
print(df)
```

```python
# Identifying missing values
print("Missing Values:")
print(df.isnull())
# Counting missing values
print("\nMissing Values Count:")
print(df.isnull().sum())
```

```python
# Identifying duplicate records
print("Duplicate Records:")
print(df[df.duplicated()])
```

```python
# Converting Age and Marks into numeric values
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')
df['Marks'] = pd.to_numeric(df['Marks'], errors='coerce')
print("Corrected Data Types:")
print(df.dtypes)
```

```python
# Filling missing values using median
df['Age'] = df['Age'].fillna(df['Age'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].median())
# Removing duplicate records
df = df.drop_duplicates()
print("Cleaned Dataset:")
print(df)
```

```python
# Final dataset information
print("Cleaned Dataset Information:")
print(df.info())
print("\nFinal Missing Values:")
print(df.isnull().sum())
print("\nFinal Duplicate Count:")
print(df.duplicated().sum())
print("\nFinal Dataset Shape:")
print(df.shape)
# Save cleaned dataset
df.to_csv("student_data_cleaned.csv", index=False)
```
