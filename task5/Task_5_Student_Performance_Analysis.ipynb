```python

import pandas as pd
import matplotlib.pyplot as plt
```

```python

df = pd.read_csv("student_data_cleaned.csv")

print("Student Performance Dataset:")
print(df)
```

```python

print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nStatistical Information:")
print(df.describe())
```

```python

print("Student Names:")
print(df['Name'])

print("\nStudent Marks:")
print(df['Marks'])

print("\nStudent Subjects:")
print(df['Subject'])
```

```python

print("Missing Values:")
print(df.isnull().sum())
```

```python

df['Age'] = df['Age'].fillna(df['Age'].median())
df['Marks'] = df['Marks'].fillna(df['Marks'].median())

print("Missing values handled successfully!")
```

```python

df = df.drop_duplicates()

print("Cleaned Dataset:")
print(df)

print("\nRemaining Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

```

```python

average_marks = df['Marks'].mean()
highest_marks = df['Marks'].max()
lowest_marks = df['Marks'].min()

print("Average Marks:", average_marks)
print("Highest Marks:", highest_marks)
print("Lowest Marks:", lowest_marks)
```

```python

pass_marks = 40

passed_students = df[df['Marks'] >= pass_marks]
failed_students = df[df['Marks'] < pass_marks]

print("Number of Students Passed:", len(passed_students))
print("Number of Students Failed:", len(failed_students))

print("\nPassed Students:")
print(passed_students[['Name', 'Marks']])

print("\nFailed Students:")
print(failed_students[['Name', 'Marks']])
```

```python

plt.figure(figsize=(8, 5))

plt.bar(df['Name'], df['Marks'])

plt.title("Student Performance Analysis")
plt.xlabel("Student Name")
plt.ylabel("Marks")

plt.show()
```

```python

plt.figure(figsize=(8, 5))

plt.hist(df['Marks'], bins=5, edgecolor='black')

plt.title("Distribution of Student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

```python

subject_average = df.groupby('Subject')['Marks'].mean()

plt.figure(figsize=(8, 5))

plt.bar(subject_average.index, subject_average.values)

plt.title("Subject-wise Average Marks")
plt.xlabel("Subject")
plt.ylabel("Average Marks")

plt.show()
```

```python

df.to_csv("student_performance_analysis.csv", index=False)

print("Final analyzed dataset saved successfully!")
```
